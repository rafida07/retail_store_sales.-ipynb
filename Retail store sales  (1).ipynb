### 2.5 Outlier Detection

```python
# Select numerical columns
df.select_dtypes(include=['int64', 'float64']).describe()
```

####  Calculate IQR boundaries

```python
# Calculate IQR boundaries

for col in ['Price_Per_Unit', 'Quantity', 'Total_Spent']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    print(col)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print()
```

#### * Count potential outliers

```python
# Count potential outliers

for col in ['Price_Per_Unit', 'Quantity', 'Total_Spent']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(col, ":", len(outliers))
```

```python
# Inspect potential outliers in Total_spent
df.loc[df['Total_Spent'] > 377.5, ['Customer_ID', 'Total_Spent']]
```

#### Using the IQR method, 157 potential outliers were found in Total_Spent. No outliers were found in Price_Per_Unit and Quantity. The values were inspected but not removed.

### 4.1 Add New Column

#### Total spent column describe

```python
df['Total_Spent'].describe()
```

```python
# create a spenting_category type
def Spenting_category(amount):
    if amount <= 100:
        return 'Low Spent'
    elif amount <= 200:
        return 'Medium Spent'
    else:
        return 'High Spent'
```

```python
# create spending category column
df['Spenting_Category'] = df['Total_Spent'].apply(Spenting_category)
```

```python
df[['Total_Spent', 'Spenting_Category']].head(10)
```
